## create a columns age_group

```python
 #convert a numerical column into a categorical one
labels = ['Young Adult' , 'Adult' , 'Middle-aged' , 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4 , labels = labels)
```

```python
df[['age_group']]
```

## purchase freq days 

```python
# the value in previous_purchase is in text . this make the analysis harder 
frequency_mapping = {
    "Fortnightly":14,
    "Weekly":7,
    "Monthly":30,
    "Quarterly":90,
    "Bi-Weekly":14,
    "Annually":365,
    "Every 3 Months":90
}
df["purchases_frequency_days"]  = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
## any sql can use mysql , mssql code is same but value is diff
#step1 : Connect to postgreSql 
username = 'postgres' #default user
password = 'root'     #inital password
host = 'localhost'    #if run locally
port = '5432'         #default port
database = 'customer_behavior'  #db created in pgAdmin

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

#step2 : Load database into postgressql 
table_name = 'customer'
df.to_sql(table_name , engine , if_exists='replace', index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}' .")
```
